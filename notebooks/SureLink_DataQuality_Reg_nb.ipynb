```python
# Read DataQualityResults from the Silver Lakehouse
dq_df = spark.read.table("SureLink_Silver_lh.dbo.dataqualityresults")

# Check the record count
print("Total quality results:", dq_df.count())

# Display the records
display(dq_df)
```

```python
# Register a managed Delta table in the Gold Lakehouse
dq_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("DataQualityResults")

print("DataQualityResults successfully registered in Gold.")
```
